# `CBERS-04`卫星`MUX`光谱响应函数导出为`ENVI`光谱库(`SLI`)文件

```python
import io, sys, os; nb_dir = os.getcwd(); 
```

```python
import zlib; 
import sqlite3; 
```

```python
import re; 
import itertools as it, collections as coll; 
from __future__ import print_function; 
```

```python
import numpy as np, pandas as pd; 
import spectral; 
```

```python
field_names = ("wavelength", "B", "G", "R", "NIR"); 
```

```python
with io.open(
    os.path.join(nb_dir, "CB04_MUX_SRF.xlsx"), 
    mode="rb"
) as xlsx: 
    crc = zlib.crc32(xlsx.read(-1)) % 2 ** 32; 
    print("CRC32: {crc:0<#x}".format(crc=crc)); 
    srf = pd.read_excel(
        xlsx.name, "MUX", header=0, names=field_names
    ); 
```

```python
srf_arr = np.array(srf).transpose(); 
mux_srf_hdr = {
    "wavelength": srf_arr[0], 
    "wavelength units": "nanometer", 
    "spectra names": field_names[1: ]
}; 
```

```python
sli = spectral.envi.SpectralLibrary(srf_arr[1: ], mux_srf_hdr, None); 
sli.save(os.path.join(nb_dir, "CB04_MUX_SRF")); 
```
